```python
def clean_state(name):
    name = name.lower()
    name = name.replace("-", " ")
    name = name.replace("&", "and")
    return name.strip()
```

```python
url="https://raw.githubusercontent.com/PhonePe/pulse/refs/heads/master/data/aggregated/transaction/country/india/state/{S}/{y}/{q}.json"
raw=[]
for s in states:
    rp=requests.get(url.format(S=s,y=2024,q=4))
    if rp.status_code != 200:
        print("FAILED:", s, rp.status_code)
        continue
    t_count=0
    t_amount=0
    for i in rp.json()["data"]["transactionData"]:
            t_count+=(i['paymentInstruments'][0]["count"])
            t_amount+=(i['paymentInstruments'][0]["amount"])
    raw.append({"state":s,"count":t_count,"amount":t_amount})
    
df=pd.DataFrame(raw)
df
```

```python
url="https://raw.githubusercontent.com/PhonePe/pulse/refs/heads/master/data/aggregated/transaction/country/india/state/{S}/{Y}/{Q}.json"
years=[2018,2019,2020,2021,2022,2023,2024]
s="maharashtra"
raw=[]
for y in years:
    for q in range(1,5):
        rp=requests.get(url.format(S=s,Y=y,Q=q))
        if rp.status_code != 200:
            print("FAILED:", s, rp.status_code)
            continue
        t_count=0
        t_amount=0
        for i in rp.json()["data"]["transactionData"]:
                t_count+=(i['paymentInstruments'][0]["count"])
                t_amount+=(i['paymentInstruments'][0]["amount"])
        raw.append({"state":s,"year":y,"quarter":q,"count":t_count,"amount":t_amount})
    
dm=pd.DataFrame(raw)
dm
```
